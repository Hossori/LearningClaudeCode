```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python
# Claude メッセージヘルパー
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens = 1000,
        messages = messages
    )
    return message.content[0].text
```

```python
messages = []
add_user_message(messages, "Define quantum computing in one sentence in Japanese")
answer = chat(messages)
add_assistant_message(messages, answer)
add_user_message(messages, "Write another sentence in Japanese")
second_answer = chat(messages)
add_assistant_message(messages, second_answer)
for message in messages:
    print(message)
```
